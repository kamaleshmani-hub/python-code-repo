```python
#while
#function_based_program
def table(table_to_print):
    multi_strt = 1
    multi_end = 10
    tables_to_print = int(input("enter the number of tables you want to print"))
    while multi_strt <= multi_end:
        print(tables_to_print, 'x', multi_strt, '=', multi_strt * tables_to_print)
        multi_strt += 1


```

```python
table(5)
```
